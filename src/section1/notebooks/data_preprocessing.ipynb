# Handling missing values

```python
# Dropping columns with too many missing values
df.drop(columns=['End_Date', 'Start_Date'], inplace=True)
```

```python
df.head()
```

```python
# Fill missing values in other columns
df['Brief_Description_of_Data'] = df['Brief_Description_of_Data'].fillna('No description provided')
df['Data_Source'] = df['Data_Source'].fillna('Unknown')
df['Time_Span'].fillna('Unknown', inplace=True)
df['Home_Department_Division'] = df['Home_Department_Division'].fillna('Unspecified')
df['Data_Steward__Business__Name'] = df['Data_Steward__Business__Name'].fillna('Not listed')
df['Data_Custodian__Technical__Name'] = df['Data_Custodian__Technical__Name'].fillna('Not listed')
df['Link_to_Existing_Publication'] = df['Link_to_Existing_Publication'].fillna('No link available')
```

```python
df.info()
```

# Cleaning Text and Columns

```python
text_columns = [
    'Brief_Description_of_Data',
    'Data_Source',
    'Home_Department',
    'Home_Department_Division',
    'Data_Steward__Business__Name',
    'Data_Custodian__Technical__Name',
    'Data_Format',
    'Frequency_of_Data_Change',
    'Time_Span',
    'Geographic_Coverage',
    'Geographic_Granularity',
    'Link_to_Existing_Publication'
]

# Strip whitespace and remove newline characters
for col in text_columns:
    df[col] = df[col].astype(str).str.strip().str.replace(r'[\n\r]+', ' ', regex=True)
```

```python
df.head()
```

```python
df.columns
```

```python
# Standardize capitalization
df['Home_Department'] = df['Home_Department'].str.title()
df['Home_Department_Division'] = df['Home_Department_Division'].str.title()
df['Data_Source'] = df['Data_Source'].str.title()
df['Data_Format'] = df['Data_Format'].str.title()
```

```python
df.head()
```

```python
# Fix Formatting in Time_Span
def clean_time_span(span):
    if pd.isna(span) or span.lower() in ['nan', 'none']:
        return 'Not specified'
    span = span.strip()
    if re.fullmatch(r'\d{4}-', span):  # e.g., "2007-"
        return span.replace('-', ' to present')
    if re.fullmatch(r'\d{4}-\d{4}', span):  # e.g., "2010-2015"
        return span.replace('-', ' to ')
    return span  # return original if no change needed

df['Time_Span'] = df['Time_Span'].apply(clean_time_span)
```
